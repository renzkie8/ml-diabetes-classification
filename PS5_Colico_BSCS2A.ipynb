##### Test Split Training (X,Y)

```python
X_training, X_testing, Y_training, Y_testing = train_test_split(X,Y,test_size=0.3,stratify=Y, random_state= 5) # x is the data and y is the label of the variable data
```

X ->> expressed x data into two arrays(training and testing)

##### SUPPORT VECTOR MACHINE TRAINING MODEL
with Kernel = sigmoid model

```python
svm_class = svm.SVC(kernel='sigmoid')
#training the svm with kernel and sigmoid model
#svc is a learning algorithm for classification tasks
#SVM uses the sigmoid kernel to map the data of the X and y training

svm_class.fit(X_training,Y_training)
```

The sigmoid kernel computes similarity between points using the sigmoid function. It's often used in neural networks and can model non-linear decision boundaries.

EVALUATING THE RESULT OF THE TRAINED MODEL

ACCURACY OF TRAINING DATASET

```python
#ACCURACY SCORE trying to predict the acc.score for all the traning data for both (X,Y)

X_training_acc = svm_class.predict(X_training) # we are predicting all the labels for X_training
# all the predicted data model is store to the X_training_acc , that will be input to the next variable
trained_data_acc = accuracy_score(X_training_acc, Y_training)
#the X_traning_acc or stored Xtraining  or the data that been trained will be compared its accuracy to the Y training label(original label)
```

```python
print('The accuracy score of the above training data:', trained_data_acc)
```

ACCURACY OF TESTING DATASET

```python
X_testing_acc = svm_class.predict(X_testing) # we are predicting all the labels for X_testing
# all the predicted data model is store to the X_testing_acc , that will be input to the next variable
testing_data_acc = accuracy_score(X_testing_acc, Y_testing)
#the X_testing_acc or stored Xtesting  or the data that been tested will be compared its accuracy to the Y tested label(original label)
```

```python
print('The accuracy score of the above testing data:', testing_data_acc)
```

TRAINING TEST SPLIT WITH MIN AND MAX

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.5, stratify=Y, random_state=42)

print(X)
print(Y)
```

Starting to Train the Model but this time will try 'linear'

```python
classifier = svm.SVC(kernel='linear')

#training the support vector Machine Classifier
classifier.fit(X_train, Y_train)
```

#### MIN_MAX accuracy Evalution

```python
# accuracy score on the training data
X_train_prediction = classifier.predict(X_train)
training_data_accuracy = accuracy_score(X_train_prediction, Y_train)
```

```python
print('Accuracy score of the training data : ', training_data_accuracy)
```

comparing it with sigmoid kernel. Linear kernel has a higher training data score.

```python
# accuracy score on the test data
X_test_prediction = classifier.predict(X_test)
test_data_accuracy = accuracy_score(X_test_prediction, Y_test)
```

```python
print('Accuracy score of the test data : ', test_data_accuracy)
```

With these result, linear kernel method has a more higher accuracy than kernel sigmoid. These mean that in data like these where class can be separated linear is more good to used than sigmoid.

It has a more simpler and efficent method having few hyperparemeter than sigmoid kernel

However Sigmoid is still used in these problem study as these problem has a two layer of network or have a little bit complex structures, therefore in a pivotal role where there is much more complext network, Sigmoid shines best than linear.

### GRAPHIC VISUAL

##### PLOTTING VISUALIZATION

```python
# Create DataFrame for visualization
diabetes_data= np.array([
    [6, 148, 72, 35, 0, 33.6, 0.627, 50, 1],
    [1, 85, 66, 29, 0, 26.6, 0.351, 31, 0],
    [8, 183, 64, 0, 0, 23.3, 0.672, 32, 1],
    [1, 89, 66, 23, 94, 28.1, 0.167, 21, 0],
    [0, 137, 40, 35, 168, 43.1, 2.288, 33, 1],
    [5, 116, 74, 0, 0, 25.6, 0.201, 30, 0]
])
columns = ['Pregnancies', 'Glucose', 'BloodPressure', 'SkinThickness', 
           'Insulin', 'BMI', 'DiabetesPedigreeFunction', 'Age', 'Outcome']

df = pd.DataFrame(diabetes_data, columns=columns)
scaler = StandardScaler()
df.iloc[:, :-1] = scaler.fit_transform(df.iloc[:, :-1])

#pairplot
sns.pairplot(df, hue='Outcome', palette='viridis',)
plt.suptitle("Pairplot of Features Colored by Diabetic Status", y=1.02)
plt.show()

#Heatmap-plot#The strength of the relationship between features.
plt.figure(figsize=(10, 8))
corr_matrix = df.iloc[:, :-1].corr()  # Exclude 'Outcome'
sns.heatmap(corr_matrix, annot=True, fmt=".2f", cmap="cubehelix", cbar=True)
plt.title("Correlation Map between Diabetes Features")
plt.show()
```

```python
features = ['Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI'] # only continous variable are on this plot
palettes = ['Dark2', 'Set2', 'magma', 'cividis', 'Spectral'] # try multiple pallete
plt.figure(figsize=(20, 15))  # Increase plot size

for i, (feature, palette) in enumerate(zip(features, palettes), 1):
    plt.subplot(2, 3, i)
    sns.histplot(data=df, x=feature, hue='Outcome', multiple='stack', palette=palette, kde=True)
    plt.title(f'Distribution of {feature} by Diabetic Status')

plt.tight_layout()
plt.show()
```

```python
#2D Scatter Plot (Using 2 features: Glucose and BMI)

print(df.columns)
plt.figure(figsize=(8, 6))
sns.scatterplot(x='Glucose', y='BMI', hue='Outcome', data=df, palette='coolwarm', s=100, edgecolor='black')
plt.title("2D Scatter Plot of Glucose vs BMI (Colored by Outcome)")
plt.xlabel('Glucose', fontsize=14)
plt.ylabel('BMI', fontsize=14)


for i in range(df.shape[0]):
    plt.text(df['Glucose'].iloc[i], df['BMI'].iloc[i], str(df['Age'].iloc[i]), 
             fontsize=9, ha='center', va='center', color='black')

# Show plot
plt.legend(title='Outcome', title_fontsize='13', loc='upper right')  # Adjust legend title and location
plt.tight_layout()  # Adjust layout to prevent clipping
plt.show()
```

```python
#Boxplot (Distribution of Features for Diabetic vs. Non-Diabetic)
plt.figure(figsize=(12, 8))
for i, col in enumerate(df.columns[:-1]):
    plt.subplot(2, 4, i + 1)  # 2 rows, 4 columns
    sns.boxplot(x='Outcome', y=col, data=df,hue='Outcome', palette='viridis')
    plt.title(col)
plt.tight_layout()
plt.suptitle("Boxplots of Features by Diabetic Status", y=1.02)
plt.show()
```
