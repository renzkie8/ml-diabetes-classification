###### STANDARDIZATION PROCESS OF THE DATA
Before the splitting of the dataset the data were first standardized before applying any classification models to the diabetes dataset, data standardization was performed to ensure that all features are on a similar scale. In this project, standardization was applied to the continuous variables, which include Glucose, BloodPressure, SkinThickness, Insulin, BMI, and Age. The standardization process involved transforming the features such that they have a mean of 0 and a standard deviation of 1. Once the features were standardized, the dataset was ready for splitting into training and test sets. We have a total of **768 original data and out of those 537 will be use within the training dataset and 231 will be done through the testing dataset**. The performance of the model was used with SVM, Kernel and Sigmoid model. 

#### CLASSIFICATION RESULT

After standardizing the dataset, the Support Vector Machine (SVM) model was used for classification, with a sigmoid kernel. The SVM is a powerful classification algorithm that works by finding the optimal hyperplane that maximally separates the different classes in the feature space. The sigmoid kernel was chosen for its ability to handle non-linear data by transforming the input into a higher-dimensional space, where a linear decision boundary can be established.

Once the model was trained on the standardized dataset, it was evaluated on the test set, and performance metrics such as accuracy score, prediction score.It was found that **accuracy score of the training data is 68%. While the accuracy score of the  testing data: 70%**.
